```python
import re
from string import punctuation
import pandas as pd
import numpy as np
from nltk import word_tokenize, sent_tokenize
from gensim.utils import simple_preprocess
```

```python
punc = set(punctuation)
punc.discard("\'")
```

```python
with open("data\movie_lines.txt") as file:
    data = file.readlines()
len(data)
```

```python
lines = []
for line in data:
    dialoge = line.split("+++$+++")[-1].strip()
    sents = sent_tokenize(dialoge)
    lines.extend(sents)
len(lines)
```

```python
def process(text):
    #lowercasing, punc, digit, special character removal
    new_text = ""
    for i in text:
        if i not in punc and not i.isdigit():
            new_text+=i
    tokens = new_text.lower().split()
    return tokens
```

```python
proc_sents = []
for line in lines:
    x = process(line)
    proc_sents.append(x)
len(proc_sents)
```

```python
df = pd.DataFrame({'text': lines, 'processed': proc_sents})
df.to_csv("data\processed_data.csv", index = False)
```
